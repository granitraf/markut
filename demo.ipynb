# Markut demo — bull-vs-bear research debate

Thin showcase notebook: all logic lives in the `markut` package. One full
governed debate (paid API calls) for one ticker.

```python
from markut import config
from markut.agents import llm
from markut.agents.graph import build_graph
from markut.evidence.edgar import ticker_to_cik
from markut.guardrails.validate import validate_ticker

TICKER = "NVDA"
```

```python
# input guardrail layers 1-2, then a fresh token ledger and the graph
ticker = validate_ticker(TICKER)
ticker_to_cik(ticker)
llm.reset()
app = build_graph()
```

```python
final_state = app.invoke({
    "ticker": ticker, "evidence": "", "bull_case": "", "bear_case": "",
    "bull_history": [], "bear_history": [], "verdict": "", "converged": False,
    "round": 0, "max_rounds": config.DEFAULT_MAX_ROUNDS, "news_evidence": "",
    "targeted_news_evidence": "", "news_checked": False,
    "claim_verification": {}, "judge_decision": {},
    "budget_exceeded": False, "review_report": {},
})
```

```python
import textwrap

print(f"rounds: {final_state['round']}  converged: {final_state['converged']}")
print(f"review: {final_state['review_report'].get('final_status')}")
print(f"usage: {llm.TOKENS['calls']} calls, {llm.TOKENS['input'] + llm.TOKENS['output']:,} tokens")
print()
print(textwrap.fill(final_state["verdict"], 100))
```
